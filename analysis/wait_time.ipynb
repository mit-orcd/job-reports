```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
with open('../data/squeue_logs.txt', 'r', encoding='utf-8') as file:
    content = file.read()

print(content)
```

```python
df = pd.read_csv('../data/squeue_logs.txt', sep='|')
df.columns = ['record_time', 'jobid', 'status', 'submit_time']
df = df[~(df.jobid.str.contains("JOBID"))] # remove the header artifact from each cron job

def parse_submit_time(time):
    time_split = time.split("T")
    return " ".join(time_split)

df['submit_time'] = df['submit_time'].apply(lambda x: parse_submit_time(x))
df['record_time'] = pd.to_datetime(df['record_time'])
df['submit_time'] = pd.to_datetime(df['submit_time'])
```

```python
TARGET_STATUSES = ["Resources","Priority"]

# Flag rows that have the target status
def is_target_status(status):
    for target_status in TARGET_STATUSES:
        if target_status.lower() in status.lower():
            return True
    return False
df['is_target_status'] = df["status"].apply(lambda x: is_target_status(x))

# Flag rows that are still pending
most_recent_record_time = df["record_time"].max()
df['is_most_recent'] = df["record_time"] == most_recent_record_time

# For each job, find the earliest record_time where is_target_status is true
earliest_target = (
    df[df["is_target_status"]]
    .groupby("jobid")["record_time"]
    .min()
    .rename("earliest_target_time")
)

# Jobs that are in non-target state at most recent record time
still_pending = (
    df[df["is_most_recent"] & ~df["jobid"].isin(earliest_target.index)]
    .groupby("jobid")["record_time"]
    .max()
    .rename("earliest_target_time")
)

still_pending[:] = pd.NaT

# Jobs that passed through target status between cron intervals
all_job_ids = df["jobid"].unique()
job_most_recent = df.groupby("jobid")["record_time"].max()
interinterval_job_ids = set(all_job_ids) - set(earliest_target.index) - set(still_pending.index)
interinterval_jobs = job_most_recent[job_most_recent.index.isin(interinterval_job_ids)].rename("earliest_target_time")


earliest_target_combined = pd.concat([earliest_target, still_pending, interinterval_jobs])
```

```python
df_jobs = (
    df.groupby("jobid")
    .agg(submit_time=("submit_time", "first"))
    .reset_index()
)
df_jobs = df_jobs.merge(earliest_target_combined.reset_index(), on="jobid", how="left")

# Compute approximate wait time (NaN for category 2)
df_jobs["approx_wait_time"] = df_jobs["earliest_target_time"] - df_jobs["submit_time"]
```
